# Project: Optimizing the Hospital Patient Journey via Process Mining and MLOps

## Introduction and Context
This project analyzes operational flows in a hospital Emergency Department (ED) using Process Mining on event logs.
Our goal is to reconstruct the real patient journey, quantify delays, and identify bottlenecks that affect throughput and resource utilization.

In parallel, we build an end-to-end MLOps pipeline to transform raw event logs into reproducible analytics (process maps + performance metrics) and predictive signals (e.g., waiting-time estimation and next-step forecasting). The final output includes automated report generation and an interactive dashboard for exploratory analysis.

## Project Objectives
1.  **Process Visualization:** Map the actual flow of activities (admission, exams..).
2.  **Performance Analysis:** Calculate throughput times and identify bottlenecks.
3.  **Anomaly Detection (Outlier Analysis):** Identify non-compliant paths or cases with anomalous timings.
4.  **Predictive Modeling:** Develop a predictive model for waiting times and the next clinical event.

## Methodology and Tech Stack
*Language*: Python 3.x

*Data Processing*: Pandas, NumPy

*Process Mining*: PM4Py (DFG discovery, performance DFG)

*Visualization*: Matplotlib (core plots), optionally Plotly for interactive views

*Graph Rendering+: Graphviz (DFG/performance graph export)

éMachine Learning*: XGBoost (primary), + baseline models where needed

*MLOps / Engineering*:
    - Docker (reproducible runs),
	- Git + GitHub (versioning, collaboration),
	- GitHub Actions (CI checks),
	- Streamlit (dashboard),
